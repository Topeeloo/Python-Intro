```python
total = 0
def add_to_total(amount):
    global total
    total += amount

add_to_total(50)
add_to_total(30)
print("Total (global):", total)    # 80

# BETTER — pass in, return out
def add_amount(current_total, amount):
    return current_total + amount

balance = 0
balance = add_amount(balance, 50)
balance = add_amount(balance, 30)
print("Balance (clean):", balance)  # 80
# No global state — easier to test and reuse
```

```python
def set_score(score):
    if not isinstance(score, (int, float)):
        raise TypeError(f'Score must be a number, got {type(score).__name__}.')
    if score < 0 or score > 100:
        raise ValueError(f'Score {score} is out of range (0-100).')
    return score
print(set_score(85))
try:
    set_score(150)
except ValueError as e:
    print(f'Caught: {e}')

try:
    set_score('A+')
except TypeError as e:
    print(f'Caught: {e}')
```

```python
def validate_name(name):
    if not isinstance(name, str) or len(name.strip()) == 0:
        raise ValueError('Name must be a non-empty string.')
    return name.strip().title()

def register_student(name, age):
    name = validate_name(name)
    if not isinstance(age, int) or age < 5 or age > 100:
        raise ValueError(f'Age {age} is not valid for a student.')
    print(f'Registered: {name}, age {age}')

try:
    register_student('  amara  ', 20)   # valid
    register_student('', 20)            # raises ValueError
except ValueError as e:
    print(f'Registration failed: {e}')
```

```python
def calculate_discounted_total(items, discount_percent):
    assert items !=[], "Items should not be empty"
    total = sum(items)
    return total * (1 - discount_percent / 100)
print(calculate_discounted_total([10, 20, 30], 10))
```

```python

def divide(a, b):
    assert b != 0, 'Denominator must not be zero'
    assert isinstance(a, (int, float)), f'Expected number, got {type(a).__name__}'
    return a / b

print(divide(10, 2))   # 5.0

try:
    divide(10, 0)
except AssertionError as e:
    print(f'AssertionError: {e}')
```
