```python
def calculate_discount(price, discount_percent=10):
    discount = ((discount_percent / 100) * price)
    final_price = price - discount
    return final_price

print(f"new price is {calculate_discount(5000, discount_percent=10)}")
print(f"new price is {calculate_discount(5000, discount_percent=20)}")
print(f"new price is {calculate_discount(5000, discount_percent=25)}")
```

```python
total = 0
def add_to_total(amount):
    global total
    total += amount

add_to_total(50)
add_to_total(30)
print("Total (global):", total)    # 80

# BETTER — pass in, return out
def add_amount(current_total, amount):
    return current_total + amount

balance = 0
balance = add_amount(balance, 50)
balance = add_amount(balance, 30)
print("Balance (clean):", balance)  # 80
# No global state — easier to test and reuse
```
