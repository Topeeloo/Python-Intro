```python
numbers = [40, 10, 30, 20, 10]
print("original list:")
print(numbers)
numbers.sort()
print("After sort():")
print(numbers)
numbers.sort(reverse=True)
print("After sort(reverse=True)")
print(numbers)
numbers.reverse()
print("After reversed():")
print(numbers)
# count()
print("count of 20:")
print(numbers.count(20))
print("index of 30:")
print(numbers.index(30))
```
